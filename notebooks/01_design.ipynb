```python
import pandas as pd

df = pd.read_csv("../data/cookie_cats.csv")

print(df.shape)
print(df.dtypes)
print(df.head())
print(df.userid.duplicated().sum())
print(df.version.value_counts())
```

```python
from scipy.stats import chisquare

counts = df.version.value_counts().sort_index()
print(counts)

expected = [len(df) / 2, len(df) / 2]
stat, p = chisquare(counts.values, f_exp=expected)
print("chi2 =", stat, "p =", p)

print(df.groupby("version")[["retention_1", "retention_7"]].mean())
```
